```python
# 6) Electricity Bill Calculator

units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 0.5
elif units <= 200:
    bill = (100 * 0.5) + ((units - 100) * 0.75)
else:
    bill = (100 * 0.5) + (100 * 0.75) + ((units - 200) * 1)

print("Total electricity bill:", bill)
```

```python
##Leap Year Checker

year = int(input("Enter a year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap year")
        else:
            print("Not a leap year")
    else:
        print("Leap year")
else:
    print("Not a leap year")
```

```python
##Triangle Classification

a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Equilateral triangle")
    elif a == b or b == c or a == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")
else:
    print("Invalid triangle")
```

```python
##Password Strength Checker

password = input("Enter password: ")

if len(password) >= 8:
    if any(char.isdigit() for char in password):
        if any(char.isupper() for char in password):
            print("Strong password")
        else:
            print("Moderate password")
    else:
        print("Moderate password")
else:
    print("Weak password")
```

```python
##Mini Banking System

account = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter current balance: "))
transaction = input("Enter transaction (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: "))

if account == "savings":
    minimum = 1000
elif account == "current":
    minimum = 5000
else:
    minimum = None

if minimum is None:
    print("Invalid account type")
elif amount <= 0:
    print("Invalid transaction amount")
elif transaction == "deposit":
    balance = balance + amount
    print("Deposit successful")
    print("New balance: ₹", balance)
elif transaction == "withdrawal":
    if balance - amount >= minimum:
        balance = balance - amount
        print("Withdrawal successful")
        print("New balance: ₹", balance)
    else:
        print("Insufficient balance")
else:
    print("Invalid transaction type")
```
